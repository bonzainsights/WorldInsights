### Option 2: Using Pandas `read_sql` (Standard)
Using standard Pandas syntax with SQLAlchemy. This allows you to use `pd.read_sql` directly, which many data scientists prefer for consistency.

````python
# Create SQL Alchemy Engine
engine = sqlalchemy.create_engine(f'duckdb:///{db_path}')

try:
    # Standard Pandas read_sql
    df = pd.read_sql(query, engine)
    print("Loaded via pd.read_sql")
    display(df.head(2))
except Exception as e:
    print(f"Error: {e}")
````
